# Messages

##### Messages are fundamental unit of context for models in langchain. They represent input and output of models, caarying both the context and metadata needed to represent a conversation whe interatcing with LLMs.

- Role: Identifies the message type(e.g, System, User)
- Content: Represents the actual content of the message(text, image, audio, doc)
- Metadata: Optional field such as response information, message ID, and token usage.

```python
import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
model = init_chat_model(model="groq:llama-3.1-8b-instant")
```

```python
model.invoke("What is Entity resolution?")
```

```python
### Text Prompt: These are strings that are ideal for straightforward questions and answers where we don't retain converstaion history. Not used for complex task.

# Use text prompt when
# 1. You have a single standalone request
# 2. You don't need to maintain context or conversation history
# 3. Minimal code complexity

model.invoke("What is langchain?")
```

```python
### Message Prompt: You can provide a list of message to the model by providing a list of message object.

# Message Types:
# 1. System Message: This is used to set the behavior of the model.
# 2. Human Message: This is used to provide input from the user.
# 3. AI Message: This is used to provide input from the model.
# 4. Tool Message: This is used to provide input from a tool.

from langchain.messages import SystemMessage, HumanMessage, AIMessage

messages = [
    SystemMessage("You are a helpful assistant Who is expert in Poetry and Literature."),
    HumanMessage("Write a small poem on Dhruv"),
]
response = model.invoke(messages)
response.content
```

```python
system_msg = SystemMessage("You are a helpful coding Assistant")

messages = [
    system_msg,
    HumanMessage("How to create a REST API in python?")
]
response = model.invoke(messages)
print(response.content)
```

```python
# Detailed info to the LLM through system message
from langchain.messages import SystemMessage, HumanMessage

system_msg = SystemMessage("You are a senior python developer with expertise in web frameworks. Always provide coding example and explain your reasoning. Be concise, and thorough on your explanations")

messages = [
    system_msg,
    HumanMessage("How to create a REST API in python?")
]

response = model.invoke(messages)
print(response.content)
```

```python
human_msg = HumanMessage(
    content="hello!",
    name="Jarvis", # Identifying different users
    id="1234" # Unique identifier for the tracing
)
```

```python
response = model.invoke([human_msg])
response
```

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

AI_msg = AIMessage("I'd be happy to help you!")

messages = [
    SystemMessage("You are a helpful Assistant"),
    HumanMessage("Can you help me?"),
    AI_msg,
    HumanMessage("Great, What is value of pi?")
]

response = model.invoke(messages)
print(response.content)
```

```python
response.usage_metadata
```

```python
from langchain.messages import AIMessage, ToolMessage

ai_msg = AIMessage(
    content=[],
    tool_calls=[{
        "name": "get_weather",
        "args": {"location": "New York"},
        "id": "call_123"
    }]   
)

weather_result = "Sunny, 32°C"

tool_msg = ToolMessage(
    content=weather_result,
    tool_call_id="call_123"
)

messages = [
    HumanMessage("What's the weather like in New York?"),
    ai_msg,
    tool_msg
]

response = model.invoke(messages)
```
