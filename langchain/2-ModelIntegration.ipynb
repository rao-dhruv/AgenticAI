### Streaming vs Batching

```python
# Streaming: When we return the chunk of text as soon as it is generated, we can use the streaming feature. This is useful for long responses or when we want to display the output in real-time.

model.invoke("What is AI?")
```

```python
for chunk in model.stream("Why do parrot have colorful feather?"):
    print(chunk.text, end="", flush=True)
```

```python
# Batching: When we send multiple prompts to the model at once, we can use the batching feature. This is useful for processing multiple inputs in parallel and can improve throughput performance and reduce cost.

responses = model.batch([
    "What is the capital of France?",
    "What is the capital of Germany?",
    "What is the capital of Italy?"
],
config={
    'max_concurrency': 5, # Optional Parameter: Maximum number of concurrent requests to the model.
    })
```

```python
for response in responses:
    print(response)
```
