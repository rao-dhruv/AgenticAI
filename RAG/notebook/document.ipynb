```python
import os
os.makedirs("../data/samples", exist_ok=True)
```

```python
sample_text = {
    "../data/samples/sample1.txt": "This is the content of sample1.txt.",
    "../data/samples/sample2.txt": "This is the content of sample2.txt.",
}

for file_path, content in sample_text.items():
    with open(file_path, "w") as f:
        f.write(content)
print("Sample text files created successfully.✅")
```

```python
from langchain_community.document_loaders import TextLoader
loader = TextLoader("../data/samples/sample1.txt")
documents = loader.load()
print(documents)
```

```python
from langchain_community.document_loaders import DirectoryLoader
from langchain_community.document_loaders import PyPDFLoader as PDFLoader
dir_loader = DirectoryLoader(
    "../data/",
    glob="**/*.pdf",
    loader_cls=PDFLoader,
    show_progress=True
)

documents = dir_loader.load()
documents
```
